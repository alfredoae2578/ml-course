## Parte 4. pandas y matplotlib: datos de cafeterías

Responde las siguientes preguntas.

````python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
%matplotlib inline

# Asegúrate de encontrarte en el directorio donde se encuentra el archivo.
# os.chdir(r'~~')     # Reemplaza el path con el tuyo.
df = pd.read_csv('data_coffeeshop.csv', header='infer', na_values=[' '])
print(df.shape)
df.head(5)
````

**1)** Genera una tabla de frecuencias para la variable `yearOfStart` y visualízala por año.
Ordena por año.
Dibuja una gráfica de línea de 1997 a 2014. *(Pista: `plt.xlim()`)*

````python
frecuencia_anual = df['Year of Start'].value_counts().sort_index()
frecuencia_anual.plot(kind='line')
plt.xlim(1997, 2014)
plt.xlabel('Año de inicio')
plt.ylabel('Número de cafeterías')
plt.title('Cafeterías abiertas por año')
plt.show()
````

**2)** Divide los datos por el estado actual del negocio (`'In'` o `'Out'`). Después, visualiza la tendencia anual de las frecuencias de la variable `yearOfStart`.
Ordena por año.
Dibuja dos gráficas de líneas superpuestas de 1997 a 2014.
Usa el objeto `figure`.

````python
fig = plt.figure()

for estado in ['In', 'Out']:
    frecuencia = df[df['Current State'] == estado]['Year of Start'].value_counts().sort_index()
    plt.plot(frecuencia.index, frecuencia.values, label=estado)

plt.xlim(1997, 2014)
plt.xlabel('Año de inicio')
plt.ylabel('Número de cafeterías')
plt.title('Cafeterías abiertas por año, según si siguen abiertas (In) o cerraron (Out)')
plt.legend()
plt.show()
````

**3)** A partir de los resultados de 1) y 2), ¿cómo describirías la tendencia en general?

**Respuesta:** Las aperturas casi no se mueven hasta el año 2000, y de ahí para adelante se disparan, sobre todo desde 2007-2008. Ya para 2013-2014 se abren miles de cafeterías al año. Separando por estado, las que siguen abiertas (In) no paran de subir, pero las que cerraron (Out) suben hasta como 2011 y después bajan. Esa bajada seguramente es porque los negocios más recientes todavía no han tenido tiempo de cerrar.
