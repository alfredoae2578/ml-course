# Ejercicio 6 - Correlación de variables

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer

data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df["target"] = 1 - data.target
```

### a) ¿Cuáles son las correlaciones con target?

```python
correlaciones = df.corr()["target"]
print(correlaciones)
```

### b) ¿Qué característica tiene la correlación más fuerte?

```python
valores = correlaciones.drop("target")
mayor = valores.abs().idxmax()
print("Característica:", mayor)
print("Correlación:", round(valores[mayor], 4))
```

### c) ¿La correlación es positiva o negativa?

```python
if valores[mayor] > 0:
    print("Positiva")
else:
    print("Negativa")
```
